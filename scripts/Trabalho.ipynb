# 1. Tratamento preliminar dos dados

1) Importação dos dados

```python
#Leitura do arquivo
df = pd.read_csv(os.path.join(input_path,'ccidRAW0.csv'),encoding='latin1')
df.columns = df.columns.str.strip() # Limpeza pois algumas colunas vem com espaços desnecessários e geraram erros
df
```

2) Formatação das variáveis

```python
df.info()
#Todas as variáveis estão sendo lidas da maneira correta.
```

3. Criação das novas variáveis

```python
#vou substituir os NaN por zero e depois substituir os zeros por NaN.
#Para isso, vou primeiro verificar se não tem zeros 

colunas_numericas = ['TMP_CD','SAL_PSU','NH4_MGN','NO3_MGN','NO2_MGN','PT_MGP','PO4_MGP','DQO','DBO','CHLA_MGC','FLR']

print('Contagem de zeros por coluna numérica:')
print((df[colunas_numericas] == 0).sum())

print('Contagem de NaN por coluna numérica:')
print((df[colunas_numericas].isna()).sum())

# Como não tem zeros, vou substituir os NaN por zeros tranquilamente
```

```python
#transformando os NaN em zero e contanto de novo
df[colunas_numericas] = df[colunas_numericas].fillna(0)

print('Contagem de zeros por coluna numérica:')
print((df[colunas_numericas] == 0).sum())

print('Contagem de NaN por coluna numérica:')
print((df[colunas_numericas].isna()).sum())
```

3a) NOx = NO2 + NO3

```python
df['NOx'] = df['NO2_MGN'] + df['NO3_MGN']
df
```

3b) Ntot = NO2 + NO3 + NH4

```python
df['Ntot'] = df['NO2_MGN'] + df['NO3_MGN'] + df['NH4_MGN']
df
```

3c) Coluna "tempo" como composição de mês e ano

```python
# Datetime é o jeito "certo" de armazenar datas no pandas
df["tempo"] = pd.to_datetime(dict(year=df["ano"], month=df["mês"], day=1)).dt.to_period("M")
df
```

3d) Variável estação a partir do mês (dez-jan-fev : verão)

```python
estacoes = {
    6: 'inverno',   7: 'inverno',   8: 'inverno',
    9: 'primavera', 10: 'primavera', 11: 'primavera',
    12: 'verão',    1: 'verão',     2: 'verão',
    3: 'outono',    4: 'outono',    5: 'outono',
}

df['estacao'] = df['mês'].map(estacoes)
df
```
